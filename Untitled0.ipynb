```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load CSV
df = pd.read_csv("/0f484464-3cc8-4bfc-968b-b1a9fc4d4b1d (1).csv")

# Inspect first rows
df.head()
```

```python
# Define target column
target = "target_5yrs"

# Separate features and target
X = df.drop(columns=[target])
y = df[target]
```

```python
# Drop player name column (non-predictive, risk of leakage)
X = X.drop(columns=["name"])
```

```python
# Compute correlation matrix
corr_matrix = X.corr()

# Visualize correlations
plt.figure(figsize=(12,8))
sns.heatmap(corr_matrix, cmap="coolwarm", center=0)
plt.title("Feature Correlation Heatmap")
plt.show()

# Identify highly correlated pairs (threshold > 0.85)
high_corr = []
for i in corr_matrix.columns:
    for j in corr_matrix.columns:
        if i != j and abs(corr_matrix.loc[i,j]) > 0.85:
            high_corr.append((i,j,corr_matrix.loc[i,j]))
```

```python
# Create Points Per Minute (PPM)
X["points_per_minute"] = df["pts"] / df["min"]

# Create Efficiency Rating (simplified: FG% + FT% + REB + AST - TO)
X["efficiency_rating"] = (df["fg"] + df["ft"]) + (df["reb"] + df["ast"]) - df["tov"]
```

```python
# Check missing values
print(X.isnull().sum())

# Fill numeric NaNs with column mean
X = X.fillna(X.mean())
```

```python
# Combine cleaned features with target
final_df = pd.concat([X, y], axis=1)

# Save cleaned dataset
final_df.to_csv("nba_cleaned.csv", index=False)

final_df.head()
```
